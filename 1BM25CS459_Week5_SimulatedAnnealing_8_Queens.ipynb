```python

import random
import math

N = 8

# Calculate attacking queen pairs
def heuristic(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j]:
                conflicts += 1

            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Generate random initial state
def random_state():
    return [random.randint(0, N - 1) for _ in range(N)]


# Simulated Annealing Algorithm
def simulated_annealing():
    current = random_state()
    current_h = heuristic(current)

    temperature = 1000
    cooling_rate = 0.99
    min_temperature = 0.001

    while temperature > min_temperature:

        if current_h == 0:
            return current

        # Generate neighboring state
        neighbor = current[:]

        col = random.randint(0, N - 1)
        row = random.randint(0, N - 1)

        neighbor[col] = row

        neighbor_h = heuristic(neighbor)

        # Calculate difference
        delta = neighbor_h - current_h

        # Accept better state
        if delta < 0:
            current = neighbor
            current_h = neighbor_h

        # Sometimes accept worse state
        else:
            probability = math.exp(-delta / temperature)

            if random.random() < probability:
                current = neighbor
                current_h = neighbor_h

        # Reduce temperature
        temperature *= cooling_rate

    return current


# Display chessboard
def display_board(state):
    print("\nFinal Chess Board:\n")

    print("  ", end="")
    for col in range(N):
        print(col, end=" ")
    print("\n  " + "--" * N)

    for row in range(N):
        print(row, end="| ")

        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()


# Main program
solution = simulated_annealing()

print("Final Solution:", solution)
print("Heuristic Value:", heuristic(solution))

display_board(solution)

if heuristic(solution) == 0:
    print("\nSolution Found!")
else:
    print("\nSolution Not Found. Try Again!")
```
